```python
from robotMovement.calculateRobotPosition import Point
from imageRecognition.positionEstimator import np_int, edge_normal
import cv2
import numpy as np
```

```python
A = np.array((400.0, 200.0))
B = np.array((300.0, 300.0))

def mouse(event, x, y, flags, param):
	if (flags & cv2.EVENT_FLAG_LBUTTON):
		global A
		A = np.array((x, y), float)
	if (flags & cv2.EVENT_FLAG_RBUTTON):
		global B
		B = np.array((x, y), float)

while True:
	frame = np.zeros((540, 960, 3))

	cv2.drawMarker(frame, np_int(A), (0, 0.5, 0), thickness=2)
	cv2.drawMarker(frame, np_int(B), (0, 1.0, 0), thickness=2)

	AB = B - A
	if (np.linalg.norm(AB) > 0):
		AB /= np.linalg.norm(AB)
	v = np.array((-AB[1], AB[0]))
	cv2.arrowedLine(frame, np_int(A), np_int(A + AB * 50), (1, 0, 0), thickness=2)
	cv2.arrowedLine(frame, np_int(A), np_int(A + v * 50), (1, 0, 0), thickness=2)

	cv2.imshow("Test", frame)
	cv2.setMouseCallback("Test", mouse)
	if (cv2.waitKey(1) != -1):
		cv2.destroyAllWindows()
		break
```
