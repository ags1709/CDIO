```python
from robotMovement.calculateRobotPosition import calculateRobotPositionFlexible
from imageRecognition.positionEstimator import np_int
import cv2
import numpy as np

fl = np.array((300.0, 200.0))
fr = np.array((400.0, 200.0))
bl = np.array((300.0, 300.0))
br = np.array((400.0, 300.0))
fl_hidden = False
fr_hidden = False
bl_hidden = False
br_hidden = False

WIDTH = 700
HEIGHT = 500

def mouse(event, x, y, flags, param):
	if (event != cv2.EVENT_LBUTTONDOWN): return
	global fl_hidden, fr_hidden, bl_hidden, br_hidden
	if (x < WIDTH/2 and y < HEIGHT/2): fl_hidden = not fl_hidden
	if (x >= WIDTH/2 and y < HEIGHT/2): fr_hidden = not fr_hidden
	if (x < WIDTH/2 and y >= HEIGHT/2): bl_hidden = not bl_hidden
	if (x >= WIDTH/2 and y >= HEIGHT/2): br_hidden = not br_hidden

while True:
	frame = np.zeros((HEIGHT, WIDTH, 3))

	cv2.drawMarker(frame, np_int(fl), (0, 0.25, 0.0) if fl_hidden else (0, 0.5, 0), thickness=2)
	cv2.drawMarker(frame, np_int(fr), (0, 0.50, 0.0) if fr_hidden else (0, 1.0, 0), thickness=2)
	cv2.drawMarker(frame, np_int(bl), (0, 0.25, 0.5) if bl_hidden else (0, 0.5, 1), thickness=2)
	cv2.drawMarker(frame, np_int(br), (0, 0.25, 0.5) if br_hidden else (0, 1.0, 1), thickness=2)

	_f, _b, _fl, _fr, _bl, _br = calculateRobotPositionFlexible(None if fl_hidden else fl, None if fr_hidden else fr, None if bl_hidden else bl, None if br_hidden else br, 75, 125)

	if _fl is not None: cv2.drawMarker(frame, np_int(_fl), (0.5, 0.5, 0), markerType=cv2.MARKER_TILTED_CROSS, thickness=2)
	if _fr is not None: cv2.drawMarker(frame, np_int(_fr), (0.5, 1.0, 0), markerType=cv2.MARKER_TILTED_CROSS, thickness=2)
	if _bl is not None: cv2.drawMarker(frame, np_int(_bl), (0.5, 0.5, 1), markerType=cv2.MARKER_TILTED_CROSS, thickness=2)
	if _br is not None: cv2.drawMarker(frame, np_int(_br), (0.5, 1.0, 1), markerType=cv2.MARKER_TILTED_CROSS, thickness=2)

	cv2.imshow("Test", frame)
	cv2.setMouseCallback("Test", mouse)
	if (cv2.waitKey(1) != -1):
		cv2.destroyAllWindows()
		break
```
